```python
# @title 3. 初始化会话
from rag_hotpotqa import InteractiveRAGSession

SUB_SET = 5  # 测试样本数，可根据需要调整

# 创建会话
session = InteractiveRAGSession()

# 加载数据（只加载一次，后续复用缓存）
session.load_data(SUB_SET)

# 加载 Qwen 生成器
session.load_generator("qwen", model_name="qwen-plus")

# 查看状态
print(session.status())
```

```python
# @title 4. 实验 1: BM25 检索
# 代码执行:
# results_bm25 = session.run_experiment(["bm25"], top_k=3)

# CLI 等价命令:
!python rag_hotpotqa.py --generator-type qwen --model qwen-plus --retriever bm25 --subset 5 --top_k 3
```
